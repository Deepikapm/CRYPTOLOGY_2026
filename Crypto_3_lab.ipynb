```python
text = "CRYPTOLOGYISINTERESTING"

freq = {}
for ch in text:
    freq[ch] = freq.get(ch, 0) + 1

for ch, count in freq.items():
    print(ch, ":", count)

# finding most frequent character 
max_char = max(freq, key=freq.get)
print("Most frequent character:", max_char, "->", freq[max_char])
```

```python
text = "CRYPTOLOGYISINTERESTING"

total = len(text)
for ch in set(text):
    count = text.count(ch)
    print(ch, ":", count, "->", round((count/total)*100, 2), "%")

print("Most frequent:", max(set(text), key=text.count))
```

```python
# check if the no is prime
num = int(input("Enter a number: "))
if num > 1:
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            print(num, "is not prime")
            break
    else:
        print(num, "is prime")
else:
    print(num, "is not prime")

start = int(input("Enter start of range: "))
end = int(input("Enter end of range: "))

print("Primes in range:")
for n in range(start, end+1):
    if n > 1:
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                break
        else:
            print(n, end=" ")
```

```python
import math

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

g = math.gcd(a, b)
print("GCD =", g)

if g == 1:
    print("They are coprime")
else:
    print("They are not coprime")
```
